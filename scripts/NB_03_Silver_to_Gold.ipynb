```python
from pyspark.sql.functions import col

# lire commandes_silver depuis LH_Silver
df_silver = spark.read.format("delta").load(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Silver.Lakehouse/Tables/dbo/commandes_silver"
)

#lire les dimensions depuis LH_Bronze
df_clients = spark.read.format("delta").load(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Bronze.Lakehouse/Tables/dbo/clients"
)

df_produits = spark.read.format("delta").load(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Bronze.Lakehouse/Tables/dbo/produits"
)

df_magasins = spark.read.format("delta").load(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Bronze.Lakehouse/Tables/dbo/magasins"
)

df_calendrier = spark.read.format("delta").load(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Bronze.Lakehouse/Tables/dbo/calendrier"
)

# Construire la table de faits
fait_ventes = df_silver.select(
    "commande_id",
    "ligne",
    col("date_commande_date").alias("date_id"),
    "client_id",
    "produit_id",
    "magasin_id",
    "quantite",
    "prix_unitaire",
    "remise_pct",
    "montant_brut",
    "montant_net",
    "frais_livraison",
    "moyen_paiement",
    "statut_commande",
    "canal_vente"
)

# Construire les dimensions
dim_client = df_clients.select(
    "client_id",
    "prenom",
    "nom",
    "genre",
    "age",
    "ville",
    "pays",
    "segment_client"
)

dim_produit = df_produits.select(
    "produit_id",
    "nom_produit",
    "categorie",
    "marque",
    "prix_unitaire",
    "cout_achat"
)

dim_magasin = df_magasins.select(
    "magasin_id",
    "nom_magasin",
    "ville",
    "pays",
    "type"
)

dim_date = df_calendrier.select(
    "date_id",
    "annee",
    "trimestre",
    "mois_num",
    "mois_nom",
    "jour_semaine",
    "est_weekend"
)

# ecrire dans LH_Gold
fait_ventes.write.format("delta").mode("overwrite").save(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Gold.Lakehouse/Tables/dbo/fait_ventes"
)

dim_client.write.format("delta").mode("overwrite").save(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Gold.Lakehouse/Tables/dbo/dim_client"
)

dim_produit.write.format("delta").mode("overwrite").save(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Gold.Lakehouse/Tables/dbo/dim_produit"
)

dim_magasin.write.format("delta").mode("overwrite").save(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Gold.Lakehouse/Tables/dbo/dim_magasin"
)

dim_date.write.format("delta").mode("overwrite").save(
    "abfss://WS_Projet_Ryma@onelake.dfs.fabric.microsoft.com/"
    "LH_Gold.Lakehouse/Tables/dbo/dim_date"
)

display(fait_ventes)
display(dim_client)
display(dim_produit)
display(dim_magasin)
display(dim_date)
```
